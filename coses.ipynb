```python
import numpy as np
```

```python
def numder(xs, ys):
    ders = []
    for i in range(0,len(ys)-1):
        ders.append((ys[i+1]-ys[i])/(xs[i+1]-xs[i]))
    return np.array(ders)
```

```python
lamb = 1550 # nm
L = 10 # km

xs = np.array([-0.2, 0, 0.2])
dt = np.array([-33.984, 0, 34.016])

D = 1/L * numder(xs, dt)

print("D = ", round((D[0]+D[1])/2, 5), " +- 0,02 [ps / (km · nm)]")
```
